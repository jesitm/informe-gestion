Facturas Cabecera
=

- Obtengo las cabeceras de las facturas

```python
# Facturas
facturas = ['VentaID','FechaVenta','HoraVenta','SucursalNombre','VendedorNombre','ClienteNombre','MetodoPago','DescuentoVenta','TotalVenta','Año','Mes']
todas_facturas = pd.read_csv("D:/soyhenry/m3/pi/ventasTransformed.csv", encoding='UTF-8', usecols=facturas)
facturas_unicas = todas_facturas.drop_duplicates().sort_values('VentaID')
#.reset_index(drop=True)
facturas_unicas['TotalSinDescuento'] = facturas_unicas['TotalVenta']/(1-facturas_unicas['DescuentoVenta'])
facturas_unicas['TotalDescuento'] = facturas_unicas['TotalSinDescuento'] -  facturas_unicas['TotalVenta'] 
tabla_facturas = facturas_unicas[['VentaID','FechaVenta','HoraVenta','SucursalNombre','VendedorNombre','ClienteNombre','MetodoPago','DescuentoVenta','TotalVenta','TotalSinDescuento','TotalDescuento','Año','Mes']]
#print (tabla_facturas)

#convertiremos la fecha
# 1. Creamos un diccionario para traducir los meses al inglés, ya que Pandas busca por defecto en inglés
meses_es = {
    "enero": "January",
    "febrero": "February",
    "marzo": "March",
    "abril": "April",
    "mayo": "May",
    "junio": "June",
    "julio": "July",
    "agosto": "August",
    "septiembre": "September",
    "setiembre": "September",
    "octubre": "October",
    "noviembre": "November",
    "diciembre": "December",
}

# 2. Limpiamos el texto: quitamos el día de la semana (jueves, ) y los conectores "de"
# Nota: Reemplaza 'tu_columna_fecha' por el nombre real de tu columna
tabla_facturas["fecha_limpia"] = (
    tabla_facturas["FechaVenta"]
    .str.split(",")
    .str[-1]
).copy()   # Nos quedamos con " 31 de diciembre de 2015"
tabla_facturas["fecha_limpia"] = tabla_facturas["fecha_limpia"].str.replace( " de ", " ", case=False)  # Queda " 31 diciembre 2015"

# 3. Traducimos el mes al inglés usando el diccionario
for es, en in meses_es.items():
    tabla_facturas["fecha_limpia"] = tabla_facturas["fecha_limpia"].str.replace(
        es, en, case=False, regex=True
    )
# Aseguramos quitar cualquier espacio extra al inicio o al final

tabla_facturas["fecha_limpia"] = tabla_facturas["fecha_limpia"].str.strip()


# Convertir usando formato mixto para que Pandas use su motor inteligente
tabla_facturas["fecha_real"] = pd.to_datetime(
    tabla_facturas["fecha_limpia"], format="mixed", dayfirst=True
)

# Convertir al formato final que necesitas (yyyy/mm/dd)
tabla_facturas["FechaVenta"] = tabla_facturas["fecha_real"].dt.strftime("%d/%m/%Y")
tabla_facturas = tabla_facturas.drop(columns=['fecha_limpia','fecha_real'])
print (tabla_facturas)

#verificamos consistencia de los datos
tabla_facturas.info()
tabla_facturas.isnull()
#obtenemos montos totales, para comparar consistencia con totales de los detalles
total_ventas = tabla_facturas['TotalVenta'].sum()
total_sin_descuento = tabla_facturas['TotalSinDescuento'].sum()
total_descuento = tabla_facturas['TotalDescuento'].sum()
print('Total ventas',total_ventas)
print('Total ventas sin descuento',total_sin_descuento)
print('Total ventas descuento',total_descuento)

```

- Vamos a actualizar la tabla facturas, para que tengo ids que correspondan, en lugar del nombre del metodo de pago, Vendedor, Sucursal, Cliente
